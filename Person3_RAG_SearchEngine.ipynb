```python
# Ensure you have installed the package first:
# !pip install -q langchain-huggingface langchain-chroma

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

# Person 1's vector database
DB_DIR = "./Ai-College-Academic-Assistant/vector_db"

print("⌒ Loading embedding model...")

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("⌒ Loading Person 1's ChromaDB...")

vector_store = Chroma(
    persist_directory=DB_DIR,
    embedding_function=embedding_model
)

print("✅ Person 1's ChromaDB loaded successfully!")
```

```python
query = "What are the units in Object Oriented Programming?"

results = vector_store.similarity_search(query, k=3)

print(f"✅ Retrieved {len(results)} documents\n")

for i, doc in enumerate(results, 1):
    print(f"--- Result {i} ---")
    print("Source:", doc.metadata.get("source_file", "Unknown"))
    print(doc.page_content[:500])
    print()
```

```python
# Similarity threshold
# Chroma returns distance scores here, so LOWER = more similar.

SIMILARITY_THRESHOLD = 1.05

query = "What are the units in Object Oriented Programming?"

results = vector_store.similarity_search_with_score(
    query,
    k=5
)

relevant_results = [
    (doc, score)
    for doc, score in results
    if score <= SIMILARITY_THRESHOLD
]

print(f"Query: {query}")
print(f"Threshold: {SIMILARITY_THRESHOLD}")
print(f"Retrieved: {len(results)}")
print(f"Relevant after filtering: {len(relevant_results)}")

for i, (doc, score) in enumerate(relevant_results, 1):
    print(f"\n--- Relevant Result {i} ---")
    print(f"Distance: {score:.4f}")
    print(f"Source: {doc.metadata.get('source_file', 'Unknown')}")
    print(doc.page_content[:400])
```

```python
def retrieve_documents(
    query,
    k=5,
    threshold=1.05
):
    """
    Retrieve top-k documents and filter them using a distance threshold.

    Lower Chroma distance = more similar.
    """

    results = vector_store.similarity_search_with_score(
        query,
        k=k
    )

    filtered_results = [
        (doc, score)
        for doc, score in results
        if score <= threshold
    ]

    return filtered_results
```

```python
query = "What are the units in Object Oriented Programming?"

results = retrieve_documents(
    query,
    k=5,
    threshold=1.05
)

print(f"Query: {query}")
print(f"Relevant documents: {len(results)}\n")

for i, (doc, score) in enumerate(results, 1):
    print(f"--- Result {i} ---")
    print(f"Distance: {score:.4f}")
    print(f"Source: {doc.metadata.get('source_file', 'Unknown')}")
    print(doc.page_content[:500])
    print()
```

```python
def build_context(results):
    """
    Combine retrieved document chunks into one context string.
    """

    if not results:
        return ""

    context_parts = []

    for i, (doc, score) in enumerate(results, 1):
        source = doc.metadata.get("source_file", "Unknown")

        context_parts.append(
            f"[Document {i} | Source: {source} | Distance: {score:.4f}]\n"
            f"{doc.page_content}"
        )

    return "\n\n".join(context_parts)
```

```python
query = "What are the units in Object Oriented Programming?"

results = retrieve_documents(
    query,
    k=5,
    threshold=1.05
)

context = build_context(results)

print("✅ Context created!\n")
print(context)
```

```python
def create_rag_prompt(question, context):
    prompt = f"""
You are a college academic assistant.

Answer the student's question using ONLY the information provided in the context below.

Rules:
1. Use only the retrieved context.
2. Do not add information from your own knowledge.
3. If the context does not contain enough information to answer the question, say:
   "I don't know based on the available college documents."
4. Keep the answer clear and concise.
5. Do not make up facts.
6. When possible, mention the source document.

CONTEXT:
{context}

STUDENT QUESTION:
{question}

ANSWER:
"""
    return prompt
```

```python
def rag_answer(question, k=5, threshold=1.05):
    """
    Complete Person 2 RAG pipeline:
    1. Retrieve top-k documents
    2. Apply similarity threshold
    3. Build context
    4. Create grounded prompt
    5. Generate answer using Gemini
    """

    # Step 1: Retrieve documents
    results = retrieve_documents(
        question,
        k=k,
        threshold=threshold
    )

    # Step 2: Check whether relevant context was found
    if not results:
        return {
            "answer": "I don't know based on the available college documents.",
            "sources": [],
            "results": []
        }

    # Step 3: Build context
    context = build_context(results)

    # Step 4: Create RAG prompt
    prompt = create_rag_prompt(
        question,
        context
    )

    # Step 5: Generate answer
    response = llm.invoke(prompt)

    # Step 6: Collect source information
    sources = [
        {
            "source": doc.metadata.get("source_file", "Unknown"),
            "distance": round(score, 4)
        }
        for doc, score in results
    ]

    return {
        "answer": response.content,
        "sources": sources,
        "results": results
    }

print("✅ RAG pipeline function created!")
```

```python
from typing import TypedDict, List, Dict
from langgraph.graph import StateGraph, START, END

print("✅ LangGraph imported successfully!")
```

```python
class AgentState(TypedDict, total=False):
    user_input: str

    # Router
    route: str

    # RAG
    answer: str
    sources: List[Dict]

    # Study Planner
    subjects: List[str]
    exam_dates: Dict[str, str]
    study_hours: float
    study_plan: str

print("✅ AgentState created successfully!")
```

```python
def router(state):
    query = state["user_input"].lower()

    study_keywords = [
        "study plan",
        "study timetable",
        "study schedule",
        "timetable",
        "schedule",
        "exam preparation",
        "prepare for exam",
        "prepare for exams",
        "how should i study",
        "how to prepare"
    ]

    if any(keyword in query for keyword in study_keywords):
        state["route"] = "study_planner"
    else:
        state["route"] = "rag"

    return state

print("✅ Router created successfully!")
```

```python
def rag_node(state):
    question = state["user_input"]

    result = rag_answer(question)

    state["answer"] = result["answer"]
    state["sources"] = result.get("sources", [])

    return state

print("✅ RAG node connected to Person 2 s work!")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

study_llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    google_api_key=GOOGLE_API_KEY,
    temperature=0.2
)

print("✅ Study Planner Gemini loaded!")
```

```python
def study_planner_node(state):

    subjects = state.get("subjects", [])
    exam_dates = state.get("exam_dates", {})
    study_hours = state.get("study_hours", 0)

    prompt = f"""
You are a college study planning assistant.

Create a realistic study timetable for a college student.

Subjects:
{subjects}

Exam dates:
{exam_dates}

Available study hours per day:
{study_hours}

Requirements:
- Prioritize exams that are closer.
- Give appropriate time to each subject.
- Include revision.
- Include reasonable breaks.
- Do not exceed the available study hours.
- Make the timetable clear and practical.

Return only the study timetable.
"""

    response = study_llm.invoke(prompt)

    state["study_plan"] = response.content
    state["answer"] = response.content

    return state

print("✅ Study Planner node created!")
```

```python
test_state = {
    "user_input": "Create a study plan",
    "subjects": [
        "Data Structures",
        "DBMS",
        "Computer Networks",
        "Operating Systems"
    ],
    "exam_dates": {
        "Data Structures": "10 October",
        "DBMS": "13 October",
        "Computer Networks": "16 October",
        "Operating Systems": "19 October"
    },
    "study_hours": 5
}

test_result = study_planner_node(test_state)

print(test_result["answer"])
```

```python
builder = StateGraph(AgentState)

builder.add_node("router", router)
builder.add_node("rag", rag_node)
builder.add_node("study_planner", study_planner_node)

builder.add_edge(START, "router")

builder.add_conditional_edges(
    "router",
    lambda state: state["route"],
    {
        "rag": "rag",
        "study_planner": "study_planner"
    }
)

builder.add_edge("rag", END)
builder.add_edge("study_planner", END)

graph = builder.compile()

print("✅ LangGraph compiled successfully!")
```

```python
result = graph.invoke({
    "user_input": "What are the units in Object Oriented Programming?"
})

print("ROUTE:")
print(result["route"])

print("\nANSWER:")
print(result["answer"])

print("\nSOURCES:")
print(result.get("sources", []))
```
